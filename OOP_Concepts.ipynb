(OOP) - What ENCAPSULATION really means

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self._balance = 1000
        self.__pin = 12345

# Create an Account obj
acct = BankAccount('Miko Logic')
# Public Attr
print(acct.owner)

# Single underscore
print(acct._balance)

# Double underscore
print(acct.__pin)

# This is because Python Automatically Mangled the name.
# Can only be accessed by knowing the Mangled name
```
